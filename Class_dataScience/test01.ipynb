

## 1. Quantile Statistics

Quantiles tell us **how the values are distributed from the smallest to the largest**.

Suppose this column contains a measurement such as **culmen depth**.

### Minimum = 13

This is the **smallest value** in the column.

```text
Minimum = 13
```

So, the lowest observed measurement is **13**.

---

### 5th Percentile = 14.2

The **5th percentile** means approximately **5% of the observations are at or below 14.2**.

```text
5%  →  14.2
```

It helps us understand the lower end of the distribution without looking only at the absolute minimum.

---

### Q1 = 15.8

**Q1 (First Quartile)** is the **25th percentile**.

That means approximately:

> **25% of the data is ≤ 15.8**

and

> **75% of the data is ≥ 15.8**

```text
0%       25%              50%              75%       100%
|---------|----------------|----------------|----------|
         15.8
          Q1
```

---

### Median = 17.8

The **median** is the middle value when the data is sorted.

It divides the data into two halves:

```text
50% of data ≤ 17.8
50% of data ≥ 17.8
```

So:

**Median = 17.8**

---

### Q3 = 18.7

**Q3 (Third Quartile)** is the **75th percentile**.

Approximately:

> **75% of the data is ≤ 18.7**

and

> **25% of the data is ≥ 18.7**

So the middle 50% of the data lies between **15.8 and 18.7**.

---

### 95th Percentile = 19.6

Approximately **95% of observations are ≤ 19.6**.

Only about **5% are above 19.6**.

This is useful for understanding the upper end of the data.

---

### Maximum = 21.4

This is the **largest observed value**.

```text
Maximum = 21.4
```

---

### Range = 8.4

Range tells us the distance between the maximum and minimum.

$$
Range = Maximum - Minimum
$$

$$
= 21.4 - 13
$$

$$
= 8.4
$$

So the complete dataset spans **8.4 units**.

---

### IQR = 2.9

**IQR = Interquartile Range**.

It represents the spread of the **middle 50% of the data**.

Formula:

$$
IQR = Q3-Q1
$$

Here:

$$
IQR = 18.7 - 15.8 = 2.9
$$

So the middle 50% of observations are spread across **2.9 units**.

# 2. Descriptive Statistics

These statistics describe the **center, spread, and shape** of the data.

---

### Standard Deviation = 1.779

Standard deviation tells us **how spread out the values are around the mean**.

Here:

```text
Standard deviation ≈ 1.78
```

A larger standard deviation means the data is more spread out.

A smaller standard deviation means the values are closer to the mean.

For your data:

```text
Mean ≈ 17.34
Standard deviation ≈ 1.78
```

So the values typically vary around the mean by roughly **1.78 units**.

---

### Coefficient of Variation (CV) = 0.1026

CV measures the size of the standard deviation **relative to the mean**.

Formula:

$$
CV = \frac{Standard\ Deviation}{Mean}
$$

Here:

$$
CV = \frac{1.779}{17.336} \approx 0.1026
$$

As a percentage:

$$
0.1026 \times 100 \approx 10.26\%
$$

So:

**CV ≈ 10.26%**

This tells us the standard deviation is about **10.26% of the mean**.

---

### Kurtosis = -0.7956

**Kurtosis describes the shape of the distribution**, particularly how heavy or light the tails are compared with a normal distribution.

Your value:

```text
Kurtosis = -0.796
```

Because it is negative, the distribution is **flatter / lighter-tailed than a normal distribution** under the convention used by many data-analysis tools (excess kurtosis).

For a beginner class, you can simply say:

> "Kurtosis helps us understand the shape and tail behavior of a distribution."

---

### Mean = 17.335556

The **mean** is the arithmetic average.

Formula:

$$
Mean = \frac{Sum\ of\ all\ values}{Number\ of\ values}
$$

Your mean is approximately:

```text
Mean = 17.34
```

So the average measurement is about **17.34**.

---

### Median Absolute Deviation (MAD) = 1.1

MAD measures how far observations typically are from the **median**.

Your:

```text
MAD = 1.1
```

So the typical absolute distance from the median is around **1.1 units**.

MAD is also useful because it is **less sensitive to extreme values (outliers)** than standard deviation.

---

### Skewness = -0.4889

**Skewness tells us whether the distribution is symmetric or has a longer tail on one side.**

Your value:

```text
Skewness = -0.489
```

Because it is **negative**, the distribution has some **left/negative skew**.

Simple interpretation:

```text
Negative skew  ←──── distribution ────→
     longer tail on the left
```

A value close to 0 would indicate a more symmetric distribution.

---

### Sum = 17,162.2

**Sum** means adding all values together.

For example:

```text
10 + 20 + 30 = 60
```

Here the total of all observations is:

```text
Sum = 17,162.2
```

And that's consistent with the mean if there are about 990 observations:

$$
Mean = \frac{17162.2}{990} \approx 17.3356
$$

---

### Variance = 3.165

Variance measures the **average squared distance from the mean**.

Your:

```text
Variance = 3.165
```

Standard deviation is related to variance:

$$
Standard\ Deviation = \sqrt{Variance}
$$

Therefore:

$$
\sqrt{3.165} \approx 1.779
$$

That's why you see:

```text
Variance           ≈ 3.165
Standard deviation ≈ 1.779
```

### Monotonicity = Not monotonic

**Monotonicity** checks whether the values consistently move in one direction.

For example, this is increasing:

```text
1 → 2 → 3 → 4 → 5
```

This is decreasing:

```text
5 → 4 → 3 → 2 → 1
```

Both are **monotonic**.

But:

```text
1 → 3 → 2 → 5 → 4
```

goes up and down.

Therefore:

**Not monotonic**

means your data does **not consistently increase or consistently decrease**.

## 🎯 Quick classroom summary

| Statistic              | Easy meaning                                |
| ---------------------- | ------------------------------------------- |
| **Minimum**            | Smallest value                              |
| **5th percentile**     | About 5% of values are below it             |
| **Q1**                 | 25% of values are below it                  |
| **Median**             | Middle value                                |
| **Q3**                 | 75% of values are below it                  |
| **95th percentile**    | About 95% of values are below it            |
| **Maximum**            | Largest value                               |
| **Range**              | Maximum − Minimum                           |
| **IQR**                | Q3 − Q1; spread of middle 50%               |
| **Standard deviation** | How spread out values are around the mean   |
| **CV**                 | Relative variation compared with the mean   |
| **Kurtosis**           | Shape/tail behavior of distribution         |
| **Mean**               | Average                                     |
| **MAD**                | Typical distance from the median            |
| **Skewness**           | Direction/degree of asymmetry               |
| **Sum**                | Total of all values                         |
| **Variance**           | Squared measure of spread                   |
| **Monotonicity**       | Whether values consistently move up or down |

### ⭐ The most important ones for beginners

Focus first on these **6**:

**Mean → Median → Minimum/Maximum → Range → Variance → Standard Deviation**

Once students understand these, move to:

**Quartiles → Percentiles → IQR → Skewness → Kurtosis → CV**.
